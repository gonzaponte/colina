```python
from pathlib import Path
from tomllib import loads

import numpy as np
import matplotlib.pyplot as plt

from nbtools import auto_plot_style
```

```python
auto_plot_style()
```

```python
maindir = Path("../data/pcolina/")
```

# 3x3 - 0.65

```python
filename = maindir / "3x3" / "0.65" / "picture.csv"
data     = np.loadtxt(filename)
n        = int(len(data)**0.5)
data     = data.reshape(n, n)

filename = filename.parent / "run.conf"
conf     = loads(open(filename).read())
sipms    = conf["geometry"]["sipm_plane"]
extent   = (sipms["sipm_size"] + sipms["sipm_gap"]) * sipms["n_sipms_side"] / 2
extent   = (-extent, extent)*2
```

```python
plt.imshow(data, origin="lower", extent=extent)
c = plt.Circle((0,0), 16, facecolor="None", edgecolor="b")
plt.gca().add_patch(c)
plt.xticks(np.arange(-15, 16, 5))
plt.yticks(np.arange(-15, 16, 5))
plt.xlabel("x (mm)")
plt.ylabel("y (mm)")
pass
```
